Task 7: Simulate Cloud Storage Allocation

```python
# Create a function allocate_storage(size)
# It should:
# - Accept storage size
# - Return confirmation message

# Write your code here
def allocate_storage(size):
    return f"Successfully allocated {size} GB of cloud storage."

storage_status = allocate_storage(cloud_resource["Storage_GB"])

print(storage_status)
```

Task 8: Simulate Auto Scaling

```python
# Create a function auto_scale(load)
# If load > 70 → scale up
# If load < 30 → scale down
# Else → maintain

# Write your code here
def auto_scale(load):
    if load > 70:
        return "High load detected. Scaling up resources."
    elif load < 30:
        return "Low load detected. Scaling down resources."
    else:
        return "Load is within the normal range. Maintaining resources."

print("Load at 85%:", auto_scale(85))
print("Load at 20%:", auto_scale(20))
print("Load at 50%:", auto_scale(50))
```

Task 9: Simulate Cloud API Interaction

```python
# Use requests to simulate a GET request to a public API
# Example: https://jsonplaceholder.typicode.com/posts/1

# Print the response

# Write your code here
api_url = "https://jsonplaceholder.typicode.com/posts/1"

try:
    response = requests.get(api_url, timeout=10)
    response.raise_for_status()

    api_data = response.json()

    print("API Response:")
    print(json.dumps(api_data, indent=4))

except requests.RequestException as error:
    print("API request failed:", error)
```

Task 10: Cost Estimation

```python
# Create a simple cost model:
# - CPU: $10 per core
# - RAM: $5 per GB
# - Storage: $1 per GB

# Calculate total cost
# Store in variable total_cost

# Write your code here
cpu_cost = cloud_resource["CPU_cores"] * 10
ram_cost = cloud_resource["RAM_GB"] * 5
storage_cost = cloud_resource["Storage_GB"] * 1

total_cost = cpu_cost + ram_cost + storage_cost

print(f"CPU Cost: ${cpu_cost:.2f}")
print(f"RAM Cost: ${ram_cost:.2f}")
print(f"Storage Cost: ${storage_cost:.2f}")
print(f"Total Estimated Cost: ${total_cost:.2f}")
```

After completing the tasks above, answer the following questions:
1. Is cloud computing always more cost-effective than on-premise infrastructure? Explain briefly.

No, cloud computing is not always more cost-effective than on-premise infrastructure. Cloud computing reduces the need to buy and maintain physical servers, but costs can increase when resources are used continuously or are not managed properly. On-premise infrastructure may be cheaper for organisations with predictable, long-term workloads.

2. Identify one limitation of cloud computing.

One limitation of cloud computing is that it depends on a reliable internet connection. If the connection is lost, users may not be able to access cloud applications, storage or other services.

3. Identify one security or ethical risk when using cloud services.

One security risk is a data breach. If cloud systems are not properly secured, unauthorised people may access sensitive information. Organisations should use strong passwords, access controls and encryption to protect their data.
